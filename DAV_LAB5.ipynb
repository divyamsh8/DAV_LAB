```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Customer_ID": ["C101","C102","C103","C104","C105","C106","C107","C108","C109","C110"],
    "City": ["Bengaluru","Mysuru","Hubballi","Bengaluru","Mangaluru",
             "Mysuru","Hubballi","Bengaluru","Mangaluru","Mysuru"],
    "Age": [25, np.nan, 31, 28, 42, 35, np.nan, 29, 38, 27],
    "Income": [45000, 38000, np.nan, 52000, 47000, 41000, 36000, np.nan, 55000, 40000],
    "Membership": ["Gold","Silver","Gold",np.nan,"Silver",
                   "Gold","Bronze","Gold","Silver","Bronze"],
    "Purchase": [3200,2100,4100,3500,2800,3900,1800,4600,5200,2400]
}

df = pd.DataFrame(data)
df
```

```python
print("First fiv values are \n")
print(df.head)
print("last five values")
print(df.tail)
print(df.dtypes)
print("Shape : ",df.shape)
print("Size : ",df.size)
```

```python
mp= (df.isnull().sum()/len(df))
mp
```

MEAN imputation


```python
age_mean = df['Age'].mean()
print("Mean age :",age_mean)
df_mean = df.copy()
df_mean["Age"] = df_mean["Age"].fillna(age_mean)
print("\n Age after mean imputation")
print(df_mean )
```

```python
income_median = df['Income'].median()
print("Median income :",income_median)
df_median = df.copy()
df_median["Income"] = df_median["Income"].fillna(income_median)
print("\n Age after median imputation")
print(df_median['Income'] )
```

```python
weather = pd.DataFrame({
    "Day" : ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"],
    "Temperature" : [25, 28,np.nan , 27, np.nan, 32,23]
})
weather
```

```python
weather_ffill = weather.copy()
weather_ffill['Temperature'].ffill(inplace=True)
print("DataFrame after forward fill imputation for 'Temperature':")
print(weather_ffill)
```

```python
copydf=df.copy()
copydf['Age']=copydf["Age"].fillna(copydf['Age'].mean())
copydf["Income"]=copydf["Income"].fillna(copydf["Income"].median())
print(copydf[["Age","Income","Membership"]].isnull().sum())
```

```python
membership_mode = copydf['Membership'].mode()[0]
print("Mode of Membership :", membership_mode)
copydf['Membership'] = copydf['Membership'].fillna(membership_mode)
print("\nNull values after Membership imputation:")
print(copydf[['Age', 'Income', 'Membership']].isnull().sum())
```

```python
encoded_df = pd.get_dummies(copydf, columns=['City', 'Membership'],dtype =int)
print("Encoded DataFrame:")
encoded_df
```

```python
avg_purchase_by_membership = copydf.groupby('Membership')['Purchase'].mean().reset_index()

plt.figure(figsize=(8, 6))
sns.barplot(x='Membership', y='Purchase', data=avg_purchase_by_membership, palette='viridis')
plt.title('Average Purchase Amount by Membership Type')
plt.xlabel('Membership Type')
plt.ylabel('Average Purchase Amount')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
