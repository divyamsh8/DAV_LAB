```python
import numpy as np
import pandas as pd
print("Numpy Version",np.__version__)
print("Pandas Version",pd.__version__)
```

```python
marks = np.array([78,85,92,67,88])
print("Marks:",marks)
print("Type: ",type(marks))
```

```python
print("Number of dimensions: ",marks.ndim)
print("Shape:",marks.shape)
print("Size:",marks.size)
print("Data Type:",marks.dtype)
print("Item Size:",marks.itemsize)
```

```python
print("Original:",marks)
print("Add 5:",marks+5)
print("Subtract 5:",marks-5)
print("Multiply 5:",marks*5)
print("Divide 5:",marks/5)
print("Power 5:",marks**5)
print("Modulus 5:",marks%5)
```

```python
print("sum:",marks.sum())
print("Mean:",marks.mean())
print("Max:",marks.max())
print("Min:",marks.min())
print("Standard Deviation:",marks.std())
print("Variance:",marks.var())
print("Product:",marks.prod())
print("Cumulative Sum:",marks.cumsum())
print("Cumulative Product:",marks.cumprod())
print("Sort:",np.sort(marks))
print("Sort in Descending Order:",np.sort(marks)[::-1])
print("Sort in Ascending Order:",np.sort(marks))
```

```python
print("First mark:",marks[0])
print("Last mark:",marks[-1])
print("First 3 marks:",marks[:3])
print("Marks after 3rd mark:",marks[3:])
print("Marks from 2nd to 4th:",marks[1:4])
print("Marks from 2nd to 4th with step 2:",marks[1:4:2])
print("Marks from 2nd to 4th with step 3:",marks[1:4:3])
print("Marks from 2nd to 4th with step 4:",marks[1:4:4])
print("Marks from 2nd to 4th with step 5:",marks[1:4:5])
print("Marks from 2nd to 4th with step 6:",marks[1:4:6])
```

```python
student_marks = np.array([[78,85,92],[67,88,76],[90,95,89]])
print("Student Marks:\n",student_marks)
print("Shape:",student_marks.shape)
print("Size:",student_marks.size)
```

```python
print("Mean of all values:",student_marks.mean())
print("Mean of each row:",student_marks.mean(axis=1))
print("Mean of each column:",student_marks.mean(axis=0))
```
