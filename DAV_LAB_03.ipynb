```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
print("Libraries imported")
```

```python
data={"Student":["Asha","Bharat","Chitra","Deepak","Esha","Farhan",
                 "Gowri","Harish","Isha","Jatin","Kavya","Lokesh",
                 "Meera","Nikhil","Pooja","Rahul","Sneha","Tarun","Uma","Varun"],
      "Department":["CSE","AIML","CSE","ISE","AIML","CSE","ISE"
      ,"CSE","AIML","ISE","CSE","AIML",
                    "ISE","CSE","AIML","ISE","CSE","AIML","ISE","CSE"],
      "Attendance":[92,78,85,96,88,72,91,65,87,82,94,76,89,97,84,69,93,80,86,90],
      "Study_Hours":[3.5,2,3,5,4,1.5,4.5,1,3.5,2.5,5.5,2,4,6,3,1.5,5,2.5,3.5,4.5],
      "Assignment_Score":[82,68,78,94,86,62,90,55,84,73,96,70,88,98,79,60,93,74,85,91],
      "Internal_Marks":[84,70,80,95,87,65,91,58,85,75,97,72,89,99,81,63,94,76,86,92],
      "Final_Score":[86,72,82,96,88,66,93,60,86,77,98,73,90,99,83,64,95,78,87,94]}
df=pd.DataFrame(data)
display(df.head())
print("Shape:",df.shape)
print(df.dtypes)
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Final_Score'],bins=6,edgecolor='black')
plt.xlabel('Final Score')
plt.ylabel('Frequency')
plt.title('Distribution of Final Scores')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Final_Score'],bins=10,edgecolor='black')
plt.xlabel('Final Score')
plt.ylabel('Frequency')
plt.title('Distribution of Final Scores')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df['Final_Score'])
plt.xlabel('Final Score ')
plt.ylabel('Frequency')
plt.title('Distribution of Final Scores')

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df['Final_Score'],x=df['Department'])
plt.xlabel('Department')
plt.ylabel('Final Score')
plt.title('Distribution of Final Scores by Department')

plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Study_Hours'],df['Final_Score'])
plt.xlabel('Study Hours')
plt.ylabel('Final Score')
plt.title('Relationship between Study Hours and Final Score')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x=df['Study_Hours'],y=df['Final_Score'],hue=df['Department'],s=10000)
plt.xlabel('Study Hours')
plt.ylabel('Final Score')
plt.title('Relationship between Study Hours, Final Score, and Department')
plt.show()
```

```python
num_columns=df.select_dtypes(include=np.number)
sns.pairplot(num_columns)
plt.show()
```

```python
corr=num_columns.corr()
display(corr)
```

```python
plt.figure(figsize=(9,6))
sns.heatmap(corr,annot=True,fmt=".2f",cmap='plasma')
plt.show()
```
