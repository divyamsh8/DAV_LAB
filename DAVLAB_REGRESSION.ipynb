```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df=pd.read_csv('housing_clean.csv')
df.head()
```

```python
X_simple=df[['Area']]
y=df['Price']
x_tr,x_te,y_tr,y_te=train_test_split(X_simple,y,test_size=0.2,random_state=42)

model_simple=LinearRegression()
model_simple.fit(x_tr,y_tr)
y_pred=model_simple.predict(x_te)

print('SIMPLE REGRESSION')
print('RMSE:',np.sqrt(mean_squared_error(y_te,y_pred)))
print('R2:',r2_score(y_te,y_pred))
```

```python
X_simple=df[['Area','Rooms','Age','Location_Score']]
y=df['Price']

x_tr,x_te,y_tr,y_te=train_test_split(X_simple,y,test_size=0.2,random_state=42)

model_multiple=LinearRegression()
model_multiple.fit(x_tr,y_tr)
y_pred=model_multiple.predict(x_te)

print('MULTIPLE REGRESSION')
print('RMSE:',np.sqrt(mean_squared_error(y_te,y_pred)))
print('R2:',r2_score(y_te,y_pred))
```

```python
residuals= y_te-y_pred
plt.scatter(y_pred,residuals,alpha=0.5)
plt.axhline(y=0,color='r',linestyle='--')
plt.xlabel('Predicted Values')
plt.ylabel('Residuals')
plt.title('Residual Plot')
plt.show()
```
